```python
import math

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

def tanh(x):
    return math.tanh(x)

print("sigmoid(0) =", sigmoid(0))
print("tanh(0)    =", tanh(0))
```

```python
Wf, Uf, bf = 0.5, 0.1, 0


Wi, Ui, bi = 0.6, 0.2, 0


Wc, Uc, bc = 0.7, 0.3, 0


Wo, Uo, bo = 0.8, 0.4, 0


Wy, by = 4, 0
```

```python
def lstm_step(x, h_prev, c_prev):


    f = sigmoid(Wf * x + Uf * h_prev + bf)


    i = sigmoid(Wi * x + Ui * h_prev + bi)


    c_hat = tanh(Wc * x + Uc * h_prev + bc)


    c = f * c_prev + i * c_hat


    o = sigmoid(Wo * x + Uo * h_prev + bo)


    h = o * tanh(c)

    return f, i, c_hat, c, o, h
```

#Time Step t=1 (x = 1)

```python
h, c = 0, 0

f, i, c_hat, c, o, h = lstm_step(1, h, c)

print("Time Step 1  |  Input = 1")
print(f"  Forget Gate        f  = {f:.3f}")
print(f"  Input  Gate        i  = {i:.3f}")
print(f"  Candidate Cell  c_hat = {c_hat:.3f}")
print(f"  Cell  State        c  = {c:.3f}")
print(f"  Output Gate        o  = {o:.3f}")
print(f"  Hidden State       h  = {h:.3f}")
```
